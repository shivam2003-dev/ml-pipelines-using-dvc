# 10-Minute Guide: Logging in Python (for MLOps)

This notebook gives you a practical logging setup you can copy into scripts like `src/data_ingestion.py`.


## 1) Why logging (1 min)
- `print()` is hard to filter and route.
- Logging gives levels (`DEBUG`, `INFO`, `WARNING`, `ERROR`), timestamps, and output to both console and files.
- In pipelines, logs are essential for debugging failures and tracing runs.


```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s | %(name)s | %(levelname)s | %(message)s'
)

logger = logging.getLogger('quickstart')
logger.info('Pipeline started')
logger.warning('Missing optional column, using fallback')
logger.error('Failed to load one artifact')
```

## 2) Use module-level loggers (2 min)
Use `LOGGER = logging.getLogger(__name__)` in each file (like your ingestion script).


```python
import logging

LOGGER = logging.getLogger(__name__)

def load_data(path: str):
    LOGGER.info('Loading data from %s', path)
    try:
        # Example only
        if not path.endswith('.csv'):
            raise ValueError('Only CSV files are supported in this demo')
        LOGGER.info('Data load successful')
    except Exception:
        LOGGER.exception('Data load failed')
        raise

load_data('train.csv')
```

## 3) Log to both console and file (3 min)
This is the most common production pattern.


```python
import logging
from pathlib import Path

Path('logs').mkdir(exist_ok=True)

logger = logging.getLogger('mlops')
logger.setLevel(logging.INFO)
logger.handlers.clear()

fmt = logging.Formatter('%(asctime)s | %(name)s | %(levelname)s | %(message)s')

console = logging.StreamHandler()
console.setFormatter(fmt)

file_handler = logging.FileHandler('logs/pipeline.log')
file_handler.setFormatter(fmt)

logger.addHandler(console)
logger.addHandler(file_handler)

logger.info('Train step started')
logger.info('Validation accuracy: %.3f', 0.912)
```

## 4) Best practices for your repo (2 min)
- Configure logging once in `main()`.
- Use `%s` formatting in logger calls (lazy and efficient).
- Use `logger.exception(...)` inside `except` blocks to keep stack traces.
- Keep `INFO` for progress and `DEBUG` for detailed diagnostics.
- Avoid logging secrets or private data.
